```python
import pandas as pd
import numpy as np
import glob
import os

# Ruta donde están tus archivos CSV crudos
raw_path = '../data/raw/'
files = sorted(glob.glob(os.path.join(raw_path, "*.csv")))

# Cargar datasets en un diccionario por año
data = {}
for file in files:
    year = os.path.basename(file).split('.')[0]
    data[year] = pd.read_csv(file)
    print(f"✅ Cargado año {year}: {data[year].shape[0]} filas, {data[year].shape[1]} columnas")
```

```python
print("==================================================")
print("1. ANÁLISIS DE NOMBRES DE COLUMNAS POR AÑO")
print("==================================================\n")

for year, df in data.items():
    print(f"--- Nombres de columnas en {year} ---")
    print(list(df.columns))
    print()
```

```python
print("==================================================")
print("2. ANÁLISIS DE TIPOS DE DATOS POR AÑO")
print("==================================================\n")

for year, df in data.items():
    print(f"--- Tipos de datos en {year} ---")
    print(df.dtypes)
    print("-" * 40)
```

```python
print("==================================================")
print("3. ANÁLISIS DE VALORES NULOS Y DUPLICADOS")
print("==================================================\n")

for year, df in data.items():
    nulls = df.isnull().sum().sum()
    dups = df.duplicated().sum()
    print(f"--- Año {year} ---")
    print(f"Total de valores nulos (NaN): {nulls}")
    print(f"Filas duplicadas exactas: {dups}")
    
    # Detalle de nulos por columna si existen
    if nulls > 0:
        print("Columnas con nulos:")
        print(df.isnull().sum()[df.isnull().sum() > 0])
    print("-" * 40)
```

```python
print("==================================================")
print("4. RESUMEN ESTADÍSTICO PARA OUTLIERS (Min, Max, Media)")
print("==================================================\n")

for year, df in data.items():
    print(f"=== Estadísticas para {year} ===")
    display(df.describe().T[['min', 'mean', 'max']])
    print()
```

```python
import pandas as pd
import glob
import os

raw_path = '../data/raw/'

# Diccionarios de mapeo de nombres de columnas a nuestro esquema unificado
mappings = {
    '2015': {
        'Country': 'country',
        'Happiness Score': 'actual_happiness_score',
        'Economy (GDP per Capita)': 'gdp',
        'Family': 'family',
        'Health (Life Expectancy)': 'health',
        'Freedom': 'freedom',
        'Trust (Government Corruption)': 'corruption',
        'Generosity': 'generosity'
    },
    '2016': {
        'Country': 'country',
        'Happiness Score': 'actual_happiness_score',
        'Economy (GDP per Capita)': 'gdp',
        'Family': 'family',
        'Health (Life Expectancy)': 'health',
        'Freedom': 'freedom',
        'Trust (Government Corruption)': 'corruption',
        'Generosity': 'generosity'
    },
    '2017': {
        'Country': 'country',
        'Happiness.Score': 'actual_happiness_score',
        'Economy..GDP.per.Capita.': 'gdp',
        'Family': 'family',
        'Health..Life.Expectancy.': 'health',
        'Freedom': 'freedom',
        'Trust..Government.Corruption.': 'corruption',
        'Generosity': 'generosity'
    },
    '2018': {
        'Country or region': 'country',
        'Score': 'actual_happiness_score',
        'GDP per capita': 'gdp',
        'Social support': 'family',
        'Healthy life expectancy': 'health',
        'Freedom to make life choices': 'freedom',
        'Perceptions of corruption': 'corruption',
        'Generosity': 'generosity'
    },
    '2019': {
        'Country or region': 'country',
        'Score': 'actual_happiness_score',
        'GDP per capita': 'gdp',
        'Social support': 'family',
        'Healthy life expectancy': 'health',
        'Freedom to make life choices': 'freedom',
        'Perceptions of corruption': 'corruption',
        'Generosity': 'generosity'
    }
}

clean_dfs = []

# Procesar cada año
for year, mapping in mappings.items():
    file_path = os.path.join(raw_path, f"{year}.csv")
    df = pd.read_csv(file_path)
    
    # Seleccionar y renombrar las columnas estandarizadas
    df_clean = df[list(mapping.keys())].rename(columns=mapping).copy()
    
    # Agregar columna de año
    df_clean['year'] = int(year)
    
    # Tratar valor nulo en 2018 (imputación por la mediana de ese año)
    if year == '2018':
        median_corruption = df_clean['corruption'].median()
        df_clean['corruption'] = df_clean['corruption'].fillna(median_corruption)
        print(f"ℹ️ Imputado valor nulo en 'corruption' (2018) con la mediana: {median_corruption:.4f}")
        
    clean_dfs.append(df_clean)

# Concatenar todos los años en un único DataFrame unificado
final_df = pd.concat(clean_dfs, ignore_index=True)

# Reordenar columnas
target_cols = ['country', 'year', 'actual_happiness_score', 'gdp', 'family', 'health', 'freedom', 'generosity', 'corruption']
final_df = final_df[target_cols]

# Asegurar carpeta de salida y guardar CSV limpio
os.makedirs('../data/processed', exist_ok=True)
output_path = '../data/processed/cleaned_happiness.csv'
final_df.to_csv(output_path, index=False)

print(f"\n✅ Dataset unificado y limpio guardado exitosamente en: {output_path}")
print(f"📊 Dimensiones finales: {final_df.shape[0]} filas, {final_df.shape[1]} columnas")
print("\nPrimeras 5 filas del dataset limpio:")
display(final_df.head())
```
